```python
import pandas as pd
```

```python
df = pd.read_sas('/Users/karina/Documents/xplanar_placement/data/RXQ_RX_G.xpt', format='xport', encoding='utf-8')


df = df.assign(RXDDRUG=df['RXDDRUG'].str.split('; ')).explode('RXDDRUG').reset_index(drop=True)
df
```

```python
top40 = df['RXDDRUG'].value_counts().head(40).index.tolist()[1:]

top40 = ['LISINOPRIL', 'SIMVASTATIN', 'METOPROLOL', 'METFORMIN', 'LEVOTHYROXINE',
       'AMLODIPINE', 'HYDROCHLOROTHIAZIDE', 'FUROSEMIDE', 'ATORVASTATIN',
       'OMEPRAZOLE', 'ATENOLOL', 'GLIPIZIDE', 'ALBUTEROL', 'WARFARIN',
       'ASPIRIN', 'POTASSIUM CHLORIDE', 'ESOMEPRAZOLE', 'CLOPIDOGREL',
       'CARVEDILOL', 'VALSARTAN', 'LOSARTAN', 'GABAPENTIN', 'ROSUVASTATIN',
       'LOVASTATIN', 'PIOGLITAZONE', 'PRAVASTATIN', 'RANITIDINE', 'TRAMADOL',
       'IBUPROFEN', 'INSULIN GLARGINE', 'CITALOPRAM', 'ALENDRONATE',
       'ALLOPURINOL', 'MONTELUKAST', 'FEXOFENADINE', 'TAMSULOSIN', 'ZOLPIDEM',
       'SERTRALINE', 'FLUTICASONE NASAL', 'PREDNISONE']

# Keep only top 40 drugs
filtered = df[df['RXDDRUG'].isin(top40)]

# Group by column SEQN and aggregate RXDDRUG
filtered = filtered.groupby('SEQN')['RXDDRUG'].apply(list).reset_index()

# remove less than 3 drugs in combination
filtered = filtered[filtered['RXDDRUG'].apply(len) >= 3]
```

```python
# get unique values from RXDDRUG
unique_drugs = set()
for drugs in filtered['RXDDRUG']:
    unique_drugs.update(drugs)
unique_drugs = list(unique_drugs)
```
